```python
import numpy as np
import sympy as sp

x1, x2, x3, x4 = sp.symbols('x1 x2 x3 x4')

A = np.array([[1,1,2,-1],[2,3,1,1],[0,-1,3,-3],[5,7,4,1]])

Arang = np.linalg.matrix_rank(A)
print("rang= ", Arang)

# Systemet rad för rad
sp.linsolve(
    [x1+x2+2*x3-x4-2, 
     2*x1+3*x2+x3+x4-4, 
     -x2+3*x3-3*x4, 
     5*x1+7*x2+4*x3+x4-10], 
    (x1, x2, x3, x4))
```
