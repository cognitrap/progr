```python
import numpy as np
import matplotlib.pyplot as plt

t=np.linspace(0, 2*np.pi, 500)

#(x1(t),y1(t)) Bakgrundsfunktionen
x1=(np.cos(t))
y1=(np.sin(t))

#Påläggsfunktionen
x2=t
y2=0.1*np.sin(10*t)

#Derivatan Bakrundsfunktionen
dx=-1*np.sin(t)
dy=np.cos(t)
dx=np.array(dx)
dy=np.array(dy)

#Normal Bakgrundsfunktionen
#Normal blir lokala y-axeln
Nx=dy/np.sqrt((dy)*(dy)+(dx)*(dx))
Ny=-dx/np.sqrt((dy)*(dy)+(dx)*(dx))

#Ändring i koordinater
Dx=y2*Nx
Dy=y2*Ny

#Nya koordinater
x3=x1+Dx
y3=y1+Dy


plt.axis('equal')

plt.plot(x1,y1)
plt.plot(x3,y3)
plt.savefig('sinuscircle.png')
plt.show()
```
